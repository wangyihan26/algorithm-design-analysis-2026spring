根据下列递归公式, 计算下列 $T(n)$ 对应的的渐近上界。要求所求的边界尽可能的紧（tight）, 请写明步骤。

1. $T(1)=1; T(n)=T(n/4)+1$ for $n>1$
2. $T(1)=1;T(n)=3T(n/3)+n^2$ for $n>1$
3. $T(1)=1;T(n)=T(2n/3)+1$ for $n>1$
4. $T(1)=1;T(n)=5T(n/4)+n$ for $n>1$
5. $T(n)=1 \ for\ n \le 2 ; T(n)=T(\sqrt{n})+1 \ for \ n>2$

**Solution:**

**1.** $T(n)=T(n/4)+1$. Master Theorem: $a=1, b=4, f(n)=1$. $n^{\log_4 1}=n^0=1$. Since $f(n)=1=\Theta(n^0)$, Case 2 applies. $T(n)=\Theta(\log n)$.

Alternatively, by repeated substitution: $T(n)=T(n/4^k)+k$. When $n/4^k=1$, $k=\log_4 n$. So $T(n)=1+\log_4 n=\Theta(\log n)$.

**2.** $T(n)=3T(n/3)+n^2$. Master Theorem: $a=3, b=3, f(n)=n^2$. $n^{\log_3 3}=n^1=n$. $f(n)=n^2=\Omega(n^{1+\epsilon})$ for $\epsilon=1>0$. Check regularity: $3(n/3)^2=n^2/3\le c\cdot n^2$ for $c=1/3<1$. Case 3 applies. $T(n)=\Theta(n^2)$.

**3.** $T(n)=T(2n/3)+1$. Master Theorem: $a=1, b=3/2, f(n)=1$. $n^{\log_{3/2}1}=n^0=1$. $f(n)=\Theta(1)=\Theta(n^0)$. Case 2 applies. $T(n)=\Theta(\log n)$.

**4.** $T(n)=5T(n/4)+n$. Master Theorem: $a=5, b=4, f(n)=n$. $n^{\log_4 5}\approx n^{1.161}$. Since $f(n)=n=O(n^{\log_4 5-\epsilon})$ for $\epsilon\approx 0.161>0$, Case 1 applies. $T(n)=\Theta(n^{\log_4 5})$.

**5.** $T(n)=T(\sqrt{n})+1, T(1)=1$ for $n\le 2$. Let $n=2^{2^k}$, then $T(2^{2^k})=T(2^{2^{k-1}})+1$. Define $S(k)=T(2^{2^k})$. Then $S(k)=S(k-1)+1$ with $S(0)=T(2)=1$. Solving: $S(k)=k+1$. Since $n=2^{2^k}$, $k=\log_2\log_2 n$. Therefore $T(n)=\Theta(\log\log n)$.

## 问题 2

观察下方的分治算法（divide-and-conquer algorithm）的伪代码, 回答下面问题

```latex
DoSomething(A,p,r)
-----
n := r-p+1
if n=2 and A[p]>A[r] then
    swap A[p] and A[r]
else if n >= 3 then
    m = ceil(2n/3)
    DoSomething(A,p,p+m-1)
    DoSomething(A,r-m+1,r)
    DoSomething(A,p,p+m-1)  
    
---
first call: DoSomething(A,1,n)
```

note：$ceil(2n/3)=\left\lceil {2n/3} \right\rceil$；$:=$ 表示赋值, 等价于 $\to$；A是一个包含n的整数元素的数组, 

1. 写出该算法时间复杂度的递归公式, 并求解其对应的渐进表示
2. 描述一下该算法的功能, 并判断是否是最高效的解决方案
3. \* 使用python编程实现上述算法或其对应的更高效版本的算法

## Analysis

The algorithm DoSomething is a divide-and-conquer sorting algorithm similar to Stooge sort.

**Recurrence:** For $n \ge 3$, $m = \lceil 2n/3 \rceil$. Three recursive calls on subarrays of size $m \approx 2n/3$:
$T(n) = 3T(2n/3) + O(1)$

**Solving:** Master Theorem: $a=3, b=3/2, f(n)=O(1)$.
$n^{\log_{3/2}3} \approx n^{2.71}$. $f(n)=O(1) = O(n^{\log_{3/2}3-\epsilon})$. Case 1 applies.
$T(n) = \Theta(n^{\log_{3/2}3}) \approx \Theta(n^{2.71})$.

**Function:** The algorithm sorts the array A in ascending order. It recursively sorts overlapping segments, ensuring the smallest elements propagate to the beginning.

**Efficiency:** Not efficient. The optimal comparison-based sorting lower bound is $\Omega(n \log n)$. Merge Sort or Quick Sort achieve $O(n \log n)$, much better than $\Theta(n^{2.71})$.

````python
import math

def DoSomething(A, p, r):
    """
    Original DoSomething algorithm (Stooge-sort variant).
    Sorts subarray A[p:r+1] in place.
    Time complexity: Theta(n^{log_{1.5}(3)}) ~ O(n^{2.71})
    """
    n = r - p + 1
    if n == 2 and A[p] > A[r]:
        A[p], A[r] = A[r], A[p]
    elif n >= 3:
        m = math.ceil(2 * n / 3)
        DoSomething(A, p, p + m - 1)
        DoSomething(A, r - m + 1, r)
        DoSomething(A, p, p + m - 1)


def merge_sort(A):
    """
    More efficient sorting algorithm: Merge Sort.
    Time complexity: Theta(n log n)
    """
    if len(A) <= 1:
        return A
    mid = len(A) // 2
    left = merge_sort(A[:mid])
    right = merge_sort(A[mid:])
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result


if __name__ == "__main__":
    A = [5, 2, 8, 1, 9, 3, 7]
    DoSomething(A, 0, len(A) - 1)
    print("DoSomething result:", A)

    B = [5, 2, 8, 1, 9, 3, 7]
    B = merge_sort(B)
    print("Merge Sort result:", B)

# your algorithm time complexity is:
# DoSomething: Theta(n^{log_{1.5}(3)}) ~ O(n^{2.71})
# Merge Sort: Theta(n log n)
````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

## Analysis

This is the classic "Best Time to Buy and Sell Stock" problem.

**Algorithm (O(n)):**
- Iterate through the array once, tracking the minimum price seen so far (`min_price`).
- At each day, compute the potential profit if we sell at the current price (`current_price - min_price`).
- Update the maximum profit if the current profit is larger.
- Record the buy and sell days for the maximum profit.

**Time Complexity:** $O(n)$ — single pass through the array, with $O(1)$ work per element.

**Space Complexity:** $O(1)$ — only a few variables.

**For the O(n^2) version:** Check all pairs (i, j) where i < j, compute profit, track max. $O(n^2)$ time, $O(1)$ space.

````python
def max_profit_brute(profits):
    """
    O(n^2) brute force: try all buy-sell pairs.
    """
    n = len(profits)
    max_p = float('-inf')
    buy_day = sell_day = -1
    for i in range(n):
        for j in range(i + 1, n):
            profit = profits[j] - profits[i]
            if profit > max_p:
                max_p = profit
                buy_day, sell_day = i + 1, j + 1  # 1-indexed days
    return max_p, buy_day, sell_day


def max_profit_linear(profits):
    """
    O(n) algorithm: track min price seen so far.
    Returns (max_profit, buy_day, sell_day) with 1-indexed days.
    """
    n = len(profits)
    if n < 2:
        return 0, -1, -1
    min_price = profits[0]
    min_day = 1
    max_profit = float('-inf')
    buy_day = sell_day = -1
    for i in range(1, n):
        profit = profits[i] - min_price
        if profit > max_profit:
            max_profit = profit
            buy_day = min_day
            sell_day = i + 1
        if profits[i] < min_price:
            min_price = profits[i]
            min_day = i + 1
    return max_profit, buy_day, sell_day


if __name__ == "__main__":
    profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]
    p1, b1, s1 = max_profit_brute(profits)
    print(f"Brute force: max profit={p1}, buy day={b1}, sell day={s1}")

    p2, b2, s2 = max_profit_linear(profits)
    print(f"O(n): max profit={p2}, buy day={b2}, sell day={s2}")

# your algorithm time complexity is:
# max_profit_brute: O(n^2)
# max_profit_linear: O(n)
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

## Analysis

**Part 1 — Finding ONE local minimum in O(log n):**

Using binary search on the array. Given the boundary conditions $A[1] > A[2]$ and $A[n-1] < A[n]$, a local minimum is guaranteed to exist.

Algorithm:
1. Check the middle element $A[mid]$.
2. If $A[mid] < A[mid-1]$ and $A[mid] < A[mid+1]$, return $A[mid]$ as a local minimum.
3. If $A[mid-1] < A[mid]$, then by the boundary conditions, a local minimum must exist in the left half. Recurse on the left half.
4. Otherwise, $A[mid+1] < A[mid]$, and a local minimum must exist in the right half. Recurse on the right half.

**Time Complexity:** $T(n) = T(n/2) + O(1) \implies \Theta(\log n)$.

**Part 2 — Finding ALL local minima:**

A single scan through the array checking each element against its neighbors finds all local minima.

**Time Complexity:** $\Theta(n)$ — we must examine every element in the worst case since there could be $\Theta(n)$ local minima (e.g., alternating high-low pattern).

````python
def find_one_local_min(A):
    """
    Find one local minimum in O(log n) using binary search.
    A local minimum is an element smaller than both neighbors.
    Assumes A[0] > A[1] and A[n-2] < A[n-1] (boundary conditions).
    Returns the value of a local minimum.
    """
    n = len(A)
    lo, hi = 0, n - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        left_smaller = (mid == 0) or (A[mid] < A[mid - 1])
        right_smaller = (mid == n - 1) or (A[mid] < A[mid + 1])
        if left_smaller and right_smaller:
            return A[mid]
        if mid > 0 and A[mid - 1] < A[mid]:
            hi = mid - 1
        else:
            lo = mid + 1
    return A[0]


def find_all_local_min(A):
    """
    Find all local minima in O(n).
    Returns list of values of all local minima.
    """
    n = len(A)
    result = []
    for i in range(n):
        left_ok = (i == 0) or (A[i] < A[i - 1])
        right_ok = (i == n - 1) or (A[i] < A[i + 1])
        if left_ok and right_ok:
            result.append(A[i])
    return result


if __name__ == "__main__":
    A = [9, 3, 7, 2, 1, 4, 5]
    print("One local min:", find_one_local_min(A))
    print("All local min:", find_all_local_min(A))

# your algorithm time complexity is:
# find_one_local_min: O(log n)
# find_all_local_min: O(n)
````

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。
2. 建立一个包含这n个数的堆（heap）, 并且调用 k 次Extract-min 按顺序输出最小的k个数。使用往空堆中不断插入元素的方法建立堆, 分析这种方法建堆的时间复杂度, 并使用python编程实现
3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

## Analysis

**Method 1 — Sort then output first k:**
Sort the array using Heap Sort or Merge Sort ($O(n \log n)$), then output the first k elements ($O(k)$). Total: $O(n \log n)$.

**Method 2 — Build heap and extract-min k times:**
Build a min-heap by inserting n elements one by one. Each insertion takes $O(\log n)$, so building takes $O(n \log n)$. k extract-min operations take $O(k \log n)$ each. Total: $O(n \log n + k \log n)$.

Note: using Floyd's heapify (build-heap from an array) achieves $O(n)$ build time, giving $O(n + k \log n)$ total.

**Method 3 — External memory (n too large for memory):**
Use a max-heap of size k to track the k smallest values seen so far.
- For each element from the data stream:
  - If the heap has fewer than k elements, insert it.
  - Otherwise, if the element is smaller than the maximum in the heap, replace the max.
- At the end, extract and sort the k elements from the heap.

**Time Complexity:** $O(n \log k)$ — each of the n elements may trigger a heap operation of cost $O(\log k)$.
**Space Complexity:** $O(k)$ — only k elements stored in memory.

````python
import heapq


def k_smallest_sort(arr, k):
    """
    Method 1: Sort then output first k.
    Time: O(n log n)
    """
    return sorted(arr)[:k]


def k_smallest_heap(arr, k):
    """
    Method 2: Build min-heap by insertion, extract-min k times.
    Time: O(n log n + k log n) with insertion-based build.
    Using heapify: O(n + k log n)
    """
    heap = arr[:]
    heapq.heapify(heap)
    return [heapq.heappop(heap) for _ in range(min(k, len(heap)))]


def k_smallest_external(arr, k):
    """
    Method 3: External memory approach using max-heap of size k.
    Time: O(n log k), Space: O(k)
    """
    if k == 0:
        return []
    max_heap = []
    for val in arr:
        if len(max_heap) < k:
            heapq.heappush(max_heap, -val)
        elif val < -max_heap[0]:
            heapq.heapreplace(max_heap, -val)
    result = sorted(-x for x in max_heap)
    return result


if __name__ == "__main__":
    arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
    k = 3
    print("Method 1 (sort):", k_smallest_sort(arr, k))
    print("Method 2 (heap):", k_smallest_heap(arr, k))
    print("Method 3 (external):", k_smallest_external(arr, k))

# your algorithm time complexity is:
# Method 1: O(n log n)
# Method 2: O(n + k log n) with heapify
# Method 3: O(n log k), Space O(k)
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

## Analysis

**Algorithm — Stack-based simulation:**

Use a stack to simulate planet collisions. Iterate through the array:
- If the current planet moves right (positive), push it onto the stack.
- If the current planet moves left (negative):
  - While the stack top is moving right (positive), a collision occurs:
    - If the left-moving planet (negative) has greater mass (absolute value) than the stack top, pop the stack top and continue checking.
    - If the left-moving planet has equal mass to the stack top (but since mass is the same, the right-moving one survives per the problem), actually: equal mass → the right-moving (positive) survives. So if abs(neg) <= stack_top (positive), the negative planet is destroyed; break.
    - Wait: the rule says equal mass → right-moving survives. So if abs(neg) == stack_top, right-moving survives, negative is destroyed.
    - If abs(neg) < stack_top, negative is destroyed (break).
    - If abs(neg) > stack_top, stack top is destroyed (pop), continue.
  - If the stack is empty or the stack top is also moving left (negative), push the current planet.

**Time Complexity:** $O(n)$ — each planet is pushed and popped at most once.
**Space Complexity:** $O(n)$ for the stack.

````python
def planet_collision(planets):
    """
    Simulate planet collisions using a stack.
    Positive = moving right, Negative = moving left.
    Equal mass + collision: right-moving (positive) survives.
    Time: O(n), Space: O(n)
    """
    stack = []
    for p in planets:
        if p > 0:
            stack.append(p)
        else:
            while stack and stack[-1] > 0:
                top = stack[-1]
                if abs(p) > top:
                    stack.pop()
                elif abs(p) == top:
                    stack.pop()
                    break
                else:
                    break
            else:
                stack.append(p)
    return stack


if __name__ == "__main__":
    A1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
    print("Test 1:", planet_collision(A1))

    A2 = [23, -8, 9, -3, -7, 9, -23, 22]
    print("Test 2:", planet_collision(A2))

# your algorithm time complexity is:
# O(n) time, O(n) space
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

## Analysis

**Algorithm — Dynamic Programming (0-1 Knapsack):**

Define $dp[i][w]$ = maximum value achievable using the first $i$ items with a weight capacity of $w$.

Recurrence:
$dp[i][w] = \max(dp[i-1][w], dp[i-1][w - weight_i] + value_i)$
where the second option is only valid if $w \ge weight_i$.

Base case: $dp[0][w] = 0$ for all $w$.

The answer is $dp[n][W]$ where $W$ is the maximum capacity.

**Time Complexity:** $O(n \cdot W)$ where $n$ is the number of items and $W$ is the capacity.
**Space Complexity:** $O(n \cdot W)$ for the DP table, or $O(W)$ with space optimization using a 1D array.

For the given example with 4 items and W=9: the optimal solution uses items (4,40) and (3,50) for a total value of 90.

````python
def knapsack_01(items, capacity):
    """
    0-1 Knapsack DP solution.
    items: list of (weight, value) tuples.
    capacity: maximum weight capacity.
    Returns the maximum value achievable.
    Time: O(n * W), Space: O(W)
    """
    dp = [0] * (capacity + 1)
    for weight, value in items:
        for w in range(capacity, weight - 1, -1):
            dp[w] = max(dp[w], dp[w - weight] + value)
    return dp[capacity]


def knapsack_with_selection(items, capacity):
    """
    Returns (max_value, selected_indices).
    Time: O(n * W), Space: O(n * W)
    """
    n = len(items)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]
    for i in range(1, n + 1):
        w_i, v_i = items[i - 1]
        for w in range(capacity + 1):
            if w_i <= w:
                dp[i][w] = max(dp[i - 1][w], dp[i - 1][w - w_i] + v_i)
            else:
                dp[i][w] = dp[i - 1][w]

    w = capacity
    selected = []
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i - 1][w]:
            selected.append(i - 1)
            w -= items[i - 1][0]

    return dp[n][capacity], selected[::-1]


if __name__ == "__main__":
    items = [(5, 10), (4, 40), (6, 30), (3, 50)]
    capacity = 9
    max_val, selected = knapsack_with_selection(items, capacity)
    print(f"Max value: {max_val}")
    print(f"Selected items (0-indexed): {selected}")
    selected_items = [items[i] for i in selected]
    print(f"Selected (weight, value): {selected_items}")

# your algorithm time complexity is:
# O(n * W) time, O(W) space (1D optimized)
````

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

## Analysis

**Algorithm — Inorder Traversal to find swapped nodes:**

In a valid BST, the inorder traversal yields elements in strictly ascending order (assuming no duplicates, or duplicates handled properly). When exactly two nodes are swapped:
- The inorder traversal will have exactly two points where the order is violated.
- The two swapped nodes are: the first element of the first violation, and the second element of the second violation (or the second element of the first violation if only one adjacent pair is violated).

Algorithm:
1. Perform an inorder traversal, collecting nodes in a list.
2. Find the two nodes that are out of order:
   - `first` = first node where `node.val > next.val`
   - `last` = last node where `prev.val > node.val`
3. Swap the values of `first` and `last`.

**Time Complexity:** $O(n)$ — one inorder traversal + one pass through the list.
**Space Complexity:** $O(n)$ for storing the traversal, or $O(h)$ if done recursively tracking just the violations (where $h$ is the tree height).

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def recover_tree(root):
    """
    Recover a BST where exactly two nodes were swapped.
    Uses O(n) time and O(h) space (Morris traversal could achieve O(1) space).
    """
    first = last = prev = None

    def inorder(node):
        nonlocal first, last, prev
        if not node:
            return
        inorder(node.left)
        if prev and prev.val > node.val:
            if not first:
                first = prev
            last = node
        prev = node
        inorder(node.right)

    inorder(root)
    if first and last:
        first.val, last.val = last.val, first.val


def tree_to_list(root):
    """Serialize tree to level-order list for display."""
    if not root:
        return []
    result = []
    q = [root]
    while q:
        node = q.pop(0)
        if node:
            result.append(node.val)
            q.append(node.left)
            q.append(node.right)
        else:
            result.append(None)
    while result and result[-1] is None:
        result.pop()
    return result


def list_to_tree(vals):
    """Build tree from level-order list."""
    if not vals:
        return None
    root = TreeNode(vals[0])
    q = [root]
    i = 1
    while q and i < len(vals):
        node = q.pop(0)
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            q.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            q.append(node.right)
        i += 1
    return root


if __name__ == "__main__":
    root1 = list_to_tree([1, 3, None, None, 2])
    print("Before:", tree_to_list(root1))
    recover_tree(root1)
    print("After:", tree_to_list(root1))

    root2 = list_to_tree([3, 1, 4, None, None, 2])
    print("Before:", tree_to_list(root2))
    recover_tree(root2)
    print("After:", tree_to_list(root2))

# your algorithm time complexity is:
# O(n) time, O(h) space
````

给定 `n` 个非负整数表示每个宽度为 `1` 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。

## Analysis

**Algorithm — Two Pointers:**

For each position i, the water it can trap is determined by the minimum of the maximum height to its left and the maximum height to its right, minus its own height: $\min(\text{left\_max}[i], \text{right\_max}[i]) - \text{height}[i]$.

Optimized two-pointer approach:
1. Use two pointers `left` and `right` starting from both ends.
2. Maintain `left_max` and `right_max` — the maximum heights seen from each side.
3. At each step, process the side with the smaller max:
   - If `height[left] < height[right]`: the water at `left` is bounded by `left_max`. Add `left_max - height[left]`, then move `left` inward.
   - Otherwise: the water at `right` is bounded by `right_max`. Add `right_max - height[right]`, then move `right` inward.

**Time Complexity:** $O(n)$ — single pass through the array.
**Space Complexity:** $O(1)$ — only a few pointers.

````python
def trap_rain_water(height):
    """
    Compute trapped rain water using two pointers.
    Time: O(n), Space: O(1)
    """
    if not height:
        return 0
    left, right = 0, len(height) - 1
    left_max = right_max = 0
    water = 0
    while left < right:
        if height[left] < height[right]:
            if height[left] >= left_max:
                left_max = height[left]
            else:
                water += left_max - height[left]
            left += 1
        else:
            if height[right] >= right_max:
                right_max = height[right]
            else:
                water += right_max - height[right]
            right -= 1
    return water


def trap_rain_water_dp(height):
    """
    DP approach using precomputed left/right max arrays.
    Time: O(n), Space: O(n)
    """
    n = len(height)
    if n == 0:
        return 0
    left_max = [0] * n
    right_max = [0] * n
    left_max[0] = height[0]
    for i in range(1, n):
        left_max[i] = max(left_max[i - 1], height[i])
    right_max[n - 1] = height[n - 1]
    for i in range(n - 2, -1, -1):
        right_max[i] = max(right_max[i + 1], height[i])
    water = 0
    for i in range(n):
        water += min(left_max[i], right_max[i]) - height[i]
    return water


if __name__ == "__main__":
    h = [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
    print("Two pointers:", trap_rain_water(h))
    print("DP:", trap_rain_water_dp(h))

# your algorithm time complexity is:
# Two pointers: O(n) time, O(1) space
# DP: O(n) time, O(n) space
````

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

## Analysis

**Algorithm — DFS traversal:**

Perform a depth-first search from the root, maintaining the current number formed by the path from root to the current node.
- At each node, update the current number: `current = current * 10 + node.val`.
- If the node is a leaf (no left or right child), add `current` to the total sum.
- Otherwise, recursively process the left and right children.

**Time Complexity:** $O(n)$ — each node is visited exactly once.
**Space Complexity:** $O(h)$ — recursion stack depth equals tree height, $O(\log n)$ for balanced trees, $O(n)$ worst case for skewed trees.

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def sum_numbers(root):
    """
    Compute sum of all root-to-leaf numbers.
    Time: O(n), Space: O(h)
    """
    def dfs(node, current_sum):
        if not node:
            return 0
        current_sum = current_sum * 10 + node.val
        if not node.left and not node.right:
            return current_sum
        return dfs(node.left, current_sum) + dfs(node.right, current_sum)

    return dfs(root, 0)


def list_to_tree(vals):
    if not vals:
        return None
    root = TreeNode(vals[0])
    q = [root]
    i = 1
    while q and i < len(vals):
        node = q.pop(0)
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            q.append(node.left)
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            q.append(node.right)
        i += 1
    return root


if __name__ == "__main__":
    root1 = list_to_tree([1, 2, 3])
    print("Test 1 sum:", sum_numbers(root1))  # 25

    root2 = list_to_tree([4, 9, 0, 5, 1])
    print("Test 2 sum:", sum_numbers(root2))  # 1026

# your algorithm time complexity is:
# O(n) time, O(h) space
````
